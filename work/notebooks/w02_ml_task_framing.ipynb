## 1) My lane as an ML task

**Lane:** Refresh / Content Opportunity Scoring

**ML task type:** Scoring

I frame this lane as a scoring task because the goal is to assign each page a content opportunity score based on available search and content signals. The score can help the SEO or content team prioritize which pages should be reviewed first.

This is more useful than treating the problem as only a yes/no decision because pages can have different levels of opportunity. A score can help create a prioritized list instead of making the same decision for every page.

## 2. Target or proxy

**Target / proxy:** A content opportunity score for each page.

Because the starter dataset does not contain a direct label showing whether a page should be refreshed, I would use a defined proxy based on observed search and content signals. The proxy would combine signals such as CTR, average position, and other available page-level features to estimate the level of content opportunity.

This would be a decision-support proxy rather than a directly observed outcome. It would help prioritize pages for review, but it would not prove that a page will improve after being refreshed.

## 3. Success metric

**Success metric:** Precision@K

I would use Precision@K to evaluate whether the highest-scored pages contain a useful share of pages that meet the defined content-opportunity criteria.

For example, if Precision@50 is 0.70, it means that 70% of the top 50 prioritized pages meet the chosen opportunity criteria. A higher Precision@K would mean the scoring approach is better at putting useful pages near the top of the review list.

The exact K value can be chosen based on how many pages the content team can realistically review in a given period.

## 5. Why ML beats a fixed rule here

A fixed rule such as "if CTR is below a certain value, refresh the page" would use only one signal and the same threshold for every page. It could miss other important factors such as average position, search volume, trend direction, engagement, content length, and search intent.

An ML-based scoring approach can consider multiple page-level signals together and learn patterns that may be difficult to capture with one fixed threshold. This can produce a more flexible priority score for deciding which pages deserve review first.

The goal is not to automatically decide that a page must be refreshed. The ML output would provide decision support for the content team, which can then review the pages and choose the appropriate action.
